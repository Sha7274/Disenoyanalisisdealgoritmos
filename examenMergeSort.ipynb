```python
from google.colab import files

def calcular_area(p1, p2, p3):
    return 0.5 * abs(p1[0] * (p2[1] - p3[1]) +
                     p2[0] * (p3[1] - p1[1]) +
                     p3[0] * (p1[1] - p2[1]))

def producto_cruz(o, a, b):
    return (a[0] - o[0]) * (b[1] - o[1]) - (a[1] - o[1]) * (b[0] - o[0])

def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2
        L = arr[:mid]
        R = arr[mid:]

        merge_sort(L)
        merge_sort(R)

        i = j = k = 0

        while i < len(L) and j < len(R):
            if L[i][0] < R[j][0] or (L[i][0] == R[j][0] and L[i][1] < R[j][1]):
                arr[k] = L[i]
                i += 1
            else:
                arr[k] = R[j]
                j += 1
            k += 1

        while i < len(L):
            arr[k] = L[i]
            i += 1
            k += 1

        while j < len(R):
            arr[k] = R[j]
            j += 1
            k += 1

    return arr

def obtener_envolvente_convexa(puntos):
    puntos_unicos = list(set(puntos))
    puntos_ordenados = merge_sort(puntos_unicos)

    if len(puntos_ordenados) <= 3:
        return puntos_ordenados

    lower = []
    for p in puntos_ordenados:
        while len(lower) >= 2 and producto_cruz(lower[-2], lower[-1], p) <= 0:
            lower.pop()
        lower.append(p)

    upper = []
    for p in reversed(puntos_ordenados):
        while len(upper) >= 2 and producto_cruz(upper[-2], upper[-1], p) <= 0:
            upper.pop()
        upper.append(p)

    return lower[:-1] + upper[:-1]

def resolver_campo_merge_sort(archivo_entrada, archivo_salida):
    puntos = []

    with open(archivo_entrada, 'r') as f:
        for linea in f:
            x, y = map(int, linea.strip().split())
            if x == -1 and y == -1:
                break
            puntos.append((x, y))

    hull = obtener_envolvente_convexa(puntos)
    h = len(hull)

    area_maxima = -1
    mejores_puntos = []

    for i in range(h):
        for j in range(i + 1, h):
            for k in range(j + 1, h):
                area_actual = calcular_area(hull[i], hull[j], hull[k])

                if area_actual > area_maxima:
                    area_maxima = area_actual
                    mejores_puntos = [hull[i], hull[j], hull[k]]

    with open(archivo_salida, 'w') as f:
        for p in mejores_puntos:
            f.write(f"{p[0]} {p[1]}\n")

    return mejores_puntos, area_maxima

mejores_puntos, area = resolver_campo_merge_sort('campo.in', 'campo.out')
files.download('campo.out')

print("Coordenadas del triángulo máximo:")
for i, punto in enumerate(mejores_puntos, 1):
    print(f"Poste {i}: X = {punto[0]}, Y = {punto[1]}")
```
