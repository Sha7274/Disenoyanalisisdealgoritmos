```python
def calcular_area(p1, p2, p3):
    return 0.5 * abs(p1[0] * (p2[1] - p3[1]) +
                     p2[0] * (p3[1] - p1[1]) +
                     p3[0] * (p1[1] - p2[1]))

def resolver_campo_fuerza_bruta(archivo_entrada, archivo_salida):
    puntos = []

    with open(archivo_entrada, 'r') as f:
        for linea in f:
            x, y = map(int, linea.strip().split())
            if x == -1 and y == -1:
                break
            puntos.append((x, y))

    n = len(puntos)
    area_maxima = -1
    mejores_puntos = []

    for i in range(n):
        for j in range(i + 1, n):
            for k in range(j + 1, n):
                area_actual = calcular_area(puntos[i], puntos[j], puntos[k])

                if area_actual > area_maxima:
                    area_maxima = area_actual
                    mejores_puntos = [puntos[i], puntos[j], puntos[k]]

    with open(archivo_salida, 'w') as f:
        for p in mejores_puntos:
            f.write(f"{p[0]} {p[1]}\n")

    return mejores_puntos, area_maxima

mejores_puntos, area = resolver_campo_fuerza_bruta('campo.in', 'campo.out')

print("Las coordenadas de los postes que generan el triángulo de mayor área son:")
for i, punto in enumerate(mejores_puntos, 1):
    print(f"Poste {i}: X = {punto[0]}, Y = {punto[1]}")
```
